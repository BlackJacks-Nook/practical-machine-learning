# Tensors

## 1. What Is A Tensor

Tabular data is commonly represented in rows and columns, allowing us to easily access, analyze, and manipulate the information.
We will encounter various tabular data structures.
- A **scalar** is the simplest form, representing a single numerical value, such as 5 or 3.14.
- A **vector** is a one-dimensional array of numbers, representing a sequence of values or a point in space, often visualized as a row or column of numbers.
- A **matrix** is a two-dimensional arrangement of numbers, consisting of multiple rows and columns.
- A **tensor** have three or more axes, making it suitable for representing complex data like images, videos, or multi-dimensional scientific data.

```python
from PIL import Image
import matplotlib.pyplot as plt

image = Image.open("../images/data/varied-data-structures.png")
plt.figure(figsize=(9, 9))
plt.imshow(image)
plt.axis("off")
plt.show()
```

This becomes more complicated because we need to carefully keep track of these different data structures and understand how they relate to one another.
In machine learning and deep learning frames, the concept of **tensor is generalized to unify all these structures (scalars, vectors, rows, and matrices) under a single, consistent framework**.

By treating data as tensors, we can handle simple values, one-dimensional sequences, two-dimensional tables, and even higher-dimensional data in a systematic way.
- A 2D tensor has the shape (rows, columns), such as a matrix of size 28×28 representing a grayscale image with 28×28 pixels.
- A 3D tensor has the shape (height, width, channels), such as a color image of 28×28 pixels with RGB channels (here, channels = 3).
- A 4D tensor has the shape (batch_size, height, width, channels), representing a dataset containing multiple color images.
- A 5D tensor has the shape (batch_size, time_steps, height, width, channels), such as a batch of 10 video clips, each consisting of 16 frames of 64×64 RGB images.
- A 6D tensor has the shape (batch_size, time_steps, depth, height, width, channels), used for 3D medical images (like MRI scans) over time: 5 patients, each with 10 time points, each scan of size 64×64×32 voxels, with 1 channel per voxel.

**Tensor** vs **ndarray**
- Tensor is similar to the fundamental object in Numpy called ndarray
- ndarray is defined as an n-dimensional homogeneous array of fixed-size items

**Advantages of Tensor**
- Tensor operations are performed significantly faster using GPUs
- Tensors can be stored and manipulated at scale using distributed processing on multi GPUs and GPUs and across multiple servers
- Tensors keep track of the graph of computations that created them
- Tensors are much more than a special sort of multi-dimensional arrays
- Interactions with each other such that transforming the tensors as a whole means that each tensor follows a particular transformation rule

## 4. Tensor Operations

### 4.1 Indexing & slicing

```python
torch.manual_seed(123456)

a_tensor = torch.rand(5,5)
print(f'a_tensor = {a_tensor} and its tensor shape = {a_tensor.shape}\n')

a_tensor_row = a_tensor[2]
print(f"A ROW of a_tensor = {a_tensor_row}\n")

a_tensor_col = a_tensor[:, 2]
print(f"A COL of a_tensor = {a_tensor_col}\n")

a_tensor_cubic = a_tensor[1:4, 1:4]
print(f"A inner cubic of a_tensor = \n{a_tensor_cubic}\n")

# use indexing to extract data that meets some criteria
print("Extra data > 0.5 = ", a_tensor_cubic > 0.5)
print("Extra data > 0.5 = ", a_tensor_cubic[a_tensor_cubic > 0.5])
```

### 4.2 Combining tensors

```python
a_tensor = torch.tensor([[0,1,2],[3,4,5]])
print(f'a_tensor = {a_tensor}\n')

b_tensor = torch.tensor([[10,11,12],[13,14,15]])
print(f'b_tensor = {b_tensor}\n')

# using `torch.stack` to add new dimension and stacks -- creates a new dimension
stack_tensor_dim0 = torch.stack((a_tensor, b_tensor), dim=0) # stack by rows
print(f'stack_tensor_dim0 = \n{stack_tensor_dim0} and its tensor shape = {stack_tensor_dim0.shape}\n')

stack_tensor_dim1 = torch.stack((a_tensor, b_tensor), dim=1) # stack by columns
print(f'stack_tensor_dim1 = \n{stack_tensor_dim1} and its tensor shape = {stack_tensor_dim1.shape}\n')

# using `torch.cat` to concatenates along existing dim -- extends an existing dimension.
cat_tensor_dim0 = torch.cat((a_tensor, b_tensor), dim=0) # cat by rows
print(f'cat_tensor_dim0 = \n{cat_tensor_dim0} and its tensor shape = {cat_tensor_dim0.shape}\n')

cat_tensor_dim1 = torch.cat((a_tensor, b_tensor), dim=1) # cat by columns
print(f'cat_tensor_dim1 = \n{cat_tensor_dim1} and its tensor shape = {cat_tensor_dim1.shape}\n')
```

### 4.3 Split tensors

| Function | Purpose | Input Parameters | Output | Changes Shape | Splits Tensor |
| :------: | :-----: | :--------------: | :------: | :------: | :------: |
| `torch.chunk` | split into fixed number of chunks | chunks (int), dim | Tuple of tensors | No | Yes |
| `torch.split` | split into chunks of specified sizes | split_size_or_sections, dim | Tuple of tensors | No | Yes |
| `torch.view` | reshape tensor | New shape | Single tensor | Yes | No |
| `torch.unbind` | remove a dimension, split into slices |dim |Tuple of tensors | Yes (removes dim) | Yes |

```python
torch.manual_seed(101)

a_tensor = torch.rand(5,5)
print(f"a_tensor = {a_tensor}\n")

print('--------- torch.unbind(tensor) ---------')
tensors = torch.unbind(a_tensor, dim=1)
print(f"first_tensor = {tensors}")

tensors = torch.unbind(a_tensor, dim=0)
print(f"first_tensor = {tensors}")

print('\n--------- torch.split(tensor) ---------')
tensors = torch.split(a_tensor, 3, dim=0) # split by rows
print(f"tensors = {tensors}\n")

tensors = torch.split(a_tensor, 3, dim=1) # split by cols
print(f"tensors = {tensors}\n")
```

```python
print('--------- torch.chunk(tensor) ---------')

chunks = torch.chunk(a_tensor, 3, dim=0)
print(f"chunks = {chunks}\n")
print(f"chunks[0] = {chunks[0]}")
print(f"chunks[1] = {chunks[1]}")
print(f"chunks[2] = {chunks[2]}\n")

chunks = torch.chunk(a_tensor, 3, dim=1)
print(f"chunks = {chunks}\n")
```

```python
# `torch.view` provides an easy way to reshape a tensor

a_tensor = torch.randn(9)
print(f"a_tensor = \n{a_tensor}\n")
b_tensor = a_tensor.view(3,3)
print(f"b_tensor = \n{b_tensor}\n")

# `torch.flatten` method can be used to collapse the dimensions of a given tensor starting with a particular dimension
a_tensor = torch.randn(2,3,4)
print(f"a_tensor = \n{a_tensor}\n")
b_tensor = torch.flatten(a_tensor, start_dim=0)
print(f"b_tensor = \n{b_tensor}\n")
c_tensor = torch.flatten(a_tensor, start_dim=1)
print(f"c_tensor = \n{c_tensor}\n")
d_tensor = torch.flatten(a_tensor, start_dim=2)
print(f"d_tensor = \n{d_tensor}\n")

print(f"tensor shapes = b:{b_tensor.shape}")
print(f"tensor shapes = c:{c_tensor.shape}")
print(f"tensor shapes = d:{d_tensor.shape}")
```

- `torch.chunk()`, it splits a tensor into a specific number of chunks, and each chunk is a view of input tensor
- `torch.split()`, it splits tensor into chunks, each chunk is a view or subdivision of original tensor
- 
- `torch.gather(), it gathers values along an axis specified by dimension

<div class='alert alert-warning'>

Some of these functions may seem redundant, however, following key distinctions and best practices are important to keep in mind
- `item()` is an important and commonly used function to return Python number from a tensor containing a single value
- use `view()` instead of `reshape()` for reshaping tensors in most cases
	- using `reshape()` may cause tensor to be copied, depending on its layout in memory
	- `view()` ensures that it will not be copied
- using `x.T` or `x.t()` is a simple way to transpose 1D or 2D tensors
	- use `transpose()` when dealing with multidimensional tensors
- `torch.squeeze()` function is used often in DL to remove an unused dimension
	- *i.e.*, a batch of images with a single image can be reduced from 4D to 3D using `squeeze()`
- `torch.unsqueeze()` function is often used in DL to add a dimension of size 1
	- since most PyTorch models expect a batch of data as an input, you could apply `unsqueeze()` when only have one data sample
	- *i.e.*, you can pass a 3D image into `torch.unsqueeze()` to create a batch of one image
</div>

#### 4.4.1 Pointwise operations

To perform an operation on each point in tensor individually and return a new tensor
- basic math functions, like `add()`, `sub()`, `mul()`, `div()`, `neg()`, and `true_divid()`
- functions for truncation, like `ceil()`, `clamp()`, `floor()`, *etc.*
- logical functions
- trigonometry functions, like `sin()`, `cos()`, *etc.*

```python
a = torch.tensor((11,12,13,14,15))
b = torch.tensor([1,2,3,4,5])

print(f"tensor a = {a}")
print(f"tensor b = {b}\n")
print(f"tensor a+b = {a+b} and {a.add(b)}")
print(f"tensor a+b = {a-b} and {a.sub(b)}")
print(f"tensor a+b = {a*b} and {a.mul(b)}")
print(f"tensor a+b = {a/b} and {a.div(b)}\n")

print(f"Sin of tensor a = {torch.sin(a)}")
print(f"Cos of tensor b = {torch.cos(b)}")
```

```python
a = torch.tensor((11,12,13))
b = torch.tensor([1,2,3])

# `dot()` allows you to compute the dot product of tensors
print(f"tensor dot = {torch.dot(a, b)}")

# `cross()` allows to compute cross product of two tensors
print(f"tensor cross = {torch.cross(a, b)}")
```

#### 4.4.2 Reduction functions

Reduce numbers down to a single number or a smaller set of numbers
- results in reducing the dimensionality or rank of a tensor
- include statistical functions such as mean, median, mode, *etc.*


```python
a = torch.tensor((11,12,13,14,15), dtype=torch.float)
b = torch.tensor([1,2,3,4,5])

print(f"Mean   of tensor a = {torch.mean(a)}")
print(f"Median of tensor b = {torch.median(b)}")
print(f"Mode   of tensor b = {torch.mode(b)}")
print(f"Standard deviation of tensor a = {torch.std(a)}")
```

<div class='alert alert-info'>

Summary of some commonly used reduction operations
- `torch.argmax(input, dim, keepdim=False, out=None)`, it returns index(es) of maximum value across all elements, or just a dimension if it’s specified
    - `torch.argmin(input, dim, keepdim=False, out=None)`, it returns index(es) of minimum value across all elements, or just a dimension if it’s specified
- `torch.dist(input, dim, keepdim=False, out=None)`, it computes p-norm of two tensors
- `torch.logsumexp(input, dim, keepdim=False, out=None)`, it computes log of summed exponentials of each row of input tensor in given dimension
- `torch.mean(input, dim, keepdim=False, out=None)`, it computes mean or average across all elements, or just a dimension if it’s specified
    - `torch.median(input, dim, keepdim=False, out=None)`, it computes median or middle value across all elements, or just a dimension if it’s specified
    - `torch.mode(input, dim, keepdim=False, out=None)`, it computes mode or most frequent value across all elements, or just a dimension if it’s specified
- `torch.norm(input, p='fro', dim=None, keepdim=False, out=None, dtype=None)`, it computes matrix or vector norm across all elements, or just a dimension if it’s specified
- `torch.prod(input, dim, keepdim=False, dtype=None)`, it computes product of all elements, or of each row of input tensor if it’s specified
- `torch.std(input, dim, keepdim=False, out=None)`, it computes standard deviation across all elements, or just a dimension if it’s specified
    - `torch.std_mean(input, unbiased=True)`, it computes standard deviation and mean across all elements, or just a dimension if it’s specified
- `torch.sum(input, dim, keepdim=False, out=None)`, it computes sum of all elements, or just a dimension if it’s specified
- `torch.unique(input, dim, keepdim=False, out=None)`, it removes duplicates across entire tensor, or just a dimension if it’s specified
	- `torch.unique_consecutive(input, dim, keepdim=False, out=None)`, it is similar to `torch.unique()` but only removes consecutive duplicates
- `torch.var(input, dim, keepdim=False, out=None)`, it computes variance across all elements, or just a dimension if it’s specified
	- `torch.var_mean(input, dim, keepdim=False, out=None)`, it computes mean and variance across all elements, or just a dimension if it’s specified
</div>

#### 4.4.3 Comparison functions

- compare all values within a tensor or compare values of two different tensors
- functions to find minimum or maximum value, sort tensor values, test tensor status or conditions, *etc.*

```python
torch.manual_seed(123456)

a_tensor = torch.rand(2,3)
b_tensor = torch.rand(2,3)
print(f"a_tensor = \n{a_tensor}\n")
print(f"b_tensor = \n{b_tensor}\n")

ge_torch = torch.ge(a_tensor, b_tensor).long()
print(f"ge_torch = \n{ge_torch}\n")

lt_torch = torch.lt(a_tensor, b_tensor).long()
print(f"lt_torch = \n{lt_torch}\n")

# torch.le(), torch.eq(), torch.ne()
```

```python
# `any` and `all` methods enable to check whether a given condition is true in any or all cases

a_tensor = torch.randn(3,3)
print(f"Tensor A = {a_tensor}\n")

print(torch.any(a > 0.0).long())
print(torch.any(a > 1.0))
print(torch.all(a > 0.0))
print(torch.all(a > 1.0))
```

<div class='alert alert-info'>

Comparison functions seem pretty straightforward, common pitfalls include following
- `torch.eq()` function or `==` returns a tensor of same size with a Boolean result for each element
	- `torch.equal()` function tests if tensors are same size, and if all elements within tensor are equal then it returns a single Boolean value

```python
# compute dot product (scalar) of two 1D tensors
a_tensor = torch.tensor([1, 2, 3, 4], dtype=torch.int32)
b_tensor = torch.tensor(np.arange(1,9,2), dtype=torch.int32)

ab_tensor = torch.matmul(a_tensor, b_tensor)
print(f"Torch.matmul for two 1D scalarss = {ab_tensor}")

# compute matrix-matrix product (2D tensor) of two 2D tensors
a_tensor = torch.tensor([[1, 2, 3], [4, 5, 6]])
b_tensor = torch.tensor([[1, 2], [3, 4], [5, 6]])

ab_tensor = torch.matmul(a_tensor, b_tensor)
print(f"Torch.matmul for two 2D matrices = {ab_tensor}\n")
```

```python
# compute a matrix product of multiple 2D tensors

a_tensor = torch.rand(2, 3)
b_tensor = torch.rand(3, 4)
c_tensor = torch.rand(4, 5)
d_tensor = torch.rand(5, 6)
abcd_tensor = torch.matmul(torch.matmul(a_tensor, b_tensor), torch.matmul(c_tensor, d_tensor))
print(f"Torch.matmul for four 2D matrices = \n{abcd_tensor}")

abcd_tensor = torch.linalg.multi_dot((a_tensor, b_tensor, c_tensor, d_tensor))
print(f"Torch.linalg.multi_dot for four 2D matrices = \n{abcd_tensor}")
```

#### 4.5.2 Automatic differentiation (Autograd)

One function, `backward()`, is worth calling out in its own subsection because it’s what makes PyTorch so powerful for DL development
- `backward()` function uses PyTorch’s automatic differentiation package, `torch.autograd`, to differentiate and compute gradients of tensors based on chain rule
- here’s a simple example of autodifferentiation
	- we define a function, `f = sum(x2)`, where $x$ is a matrix of variables
	- if we want to find $\frac{df}{dx}$ for each variable in matrix, we need to set `requires_grad = True` flag for tensor $x$
- `f.backward()` function performs differentiation with respect to $f$ and stores $\frac{df}{dx}$ in `x.grad` attribute
	- a quick review of calculus differential equations will tell us the derivation of $f$ with respect to $x$, $\frac{df}{dx}=2x$
	- results of evaluating $\frac{df}{dx}$ for values of $x$ are shown as output
- **only tensors of floating-point dtype can require gradients**

```python
x = torch.tensor([[1,2,3],[4,5,6]], dtype=torch.float, requires_grad=True)
print(x)

f = x.pow(2).sum()
print(f)

f.backward()
print(x.grad) # df/dx = 2x
```

### 5.2 Moving tensors between CPUs and GPUs

**Moving Tensors between CPUs and GPUs**
- Transfer data from CPU to GPU
- After training, output Tensors are produced in GPU
- Output data requires preprocessing
- Some  preprocessing libraries don't support Tensors and expect a Numpy array
- Numpy supports only data in GPU, and we need to move data from GPU to CPU

```python
# data from CPU to GPU --- 3 methods

a_tensor = torch.rand((2,4), device='cpu')
print(f'a_tensor = {a_tensor} and its tensor device = {a_tensor.device}\n')

# 1. `Tensor.cuda()`
b_tensor = a_tensor.cuda()
print(f'b_tensor = {b_tensor} and its tensor device = {b_tensor.device}\n')

# 2. `Tensor.to("cuda")`
b_tensor = a_tensor.to("cuda")
print(f'b_tensor = {b_tensor} and its tensor device = {b_tensor.device}\n')

# 3. `Tensor.to("cuda:0")`
b_tensor = a_tensor.to("cuda:0")
print(f'b_tensor = {b_tensor} and its tensor device = {b_tensor.device}\n')
```

```python
# data from GPU to CPU --- 2 methods

a_tensor = torch.rand((4), requires_grad=True, device='cuda')
print(f'a_tensor = {a_tensor} and its tensor device = {a_tensor.device}\n')

# 1. `Tensor.cpu()` with `requires_grad=False`
b_tensor = a_tensor.cpu()
print(f'b_tensor = {b_tensor} and its tensor device = {b_tensor.device}\n')

# 2. `Tensor.detach().cpu() with `requires_grad=True`
b_tensor = a_tensor.detach().cpu()
print(f'b_tensor = {b_tensor} and its tensor device = {b_tensor.device}\n')
```

```python
# gpu tensor -> cpu tensor -> cpu numpy array -> cpu tensor --> gpu tensor

# generate a tensor on GPU
a_tensor_gpu = torch.rand((2,2), requires_grad=True, device='cuda')
print(f'a_tensor_gpu = {a_tensor_gpu} and its tensor device = {a_tensor_gpu.device}\n')

# transfer tensor (it is still a tensor!) from GPU to CPU
a_tensor_cpu = a_tensor_gpu.detach().cpu()
print(f'a_tensor_cpu = {a_tensor_cpu} and its array device = {a_tensor_cpu.device}\n')

# convert CPU tensor to Numpy array (it is not a tensor)
a_array_from_tensor = a_tensor_cpu.numpy()
print(f'a_array_from_tensor = {a_array_from_tensor}\n') # it is not a tensor and Numpy doesnot have a `device` attribute

# convert numpy array to tensor (CPU) again 
a_tensor_from_array = torch.from_numpy(a_array_from_tensor)
print(f'a_tensor_from_array = {a_tensor_from_array} and its tensor device = {a_tensor_from_array.device}\n')
```

### 6.2 Color images

In computer vision, color images are typically represented using three channels corresponding to **Red**, **Green**, and **Blue** (RGB).
- Each channel captures the intensity of one color component for every pixel in the image.
- Converting a color image into a three-channel tensor ensures that these separate color intensities are preserved, allowing machine learning models to extract meaningful features from the combination of colors.

> Deep learning frameworks, such as PyTorch and TensorFlow, expect input images in a channel-first or channel-last tensor format.
> - PyTorch follows the channel-first (C, H, W) convention because it maps well to how CUDA/cuDNN kernels expect data for convolution operations.
> - TensorFlow/Keras defaults to channel-last (H, W, C) because it’s more natural for humans (rows × columns × colors), and it aligns with NumPy and most image processing libraries.

```python
# load a color image (replace with your image path)
image_path = "./1-berzelius.jpg"
image = Image.open(image_path).convert("RGB")  # ensure 3 channels

# show the original image
plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
import torchvision.transforms as transforms

# convert image to tensor
transform = transforms.ToTensor()
tensor_image = transform(image)  # shape in (C, H, W)

# print tensor size
print("Tensor shape:", tensor_image.shape)  # (3, H, W) means 3 channels
print(f"Min value: {tensor_image.min().item()} and Max value: {tensor_image.max().item()}")

print('\n', tensor_image)
```

```python
# split into three separate channels
red_channel = tensor_image[0:1, :, :]    # shape: (1, H, W)
green_channel = tensor_image[1:2, :, :]  # shape: (1, H, W)
blue_channel = tensor_image[2:3, :, :]   # shape: (1, H, W)
```

```python
# visualize each channel
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(red_channel.squeeze(), cmap="Reds")
axes[0].set_title("Red Channel")
axes[0].axis("off")

axes[1].imshow(green_channel.squeeze(), cmap="Greens")
axes[1].set_title("Green Channel")
axes[1].axis("off")

axes[2].imshow(blue_channel.squeeze(), cmap="Blues")
axes[2].set_title("Blue Channel")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```
